```python
n = 5

for i in range(n, 0, -1):
    for j in range(i):
        if j == 0 or j == i-1 or i == n:
            print(chr(65 + j), end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
rows = 10
for x in range(1,rows+1):
    for y in range(1, 2*rows):
        if x + y >=rows+1 and y-x <= rows-1:
            print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```
